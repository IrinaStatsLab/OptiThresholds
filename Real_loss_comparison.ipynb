# Real-data loss comparison

This notebook compares the manuscript's settled data-driven thresholds against consensus thresholds and pooled-quantile naive thresholds on the four real-data CGM cohorts: healthy, T1D, combined, and AI-READI.

Applied refinements for the revision tables:

- pooled-quantile thresholds are shown in a compact dataset-by-`K` table with reviewer-requested `K=9` deciles
- the supplementary loss table now follows the manuscript's main experiment choices only
- Healthy and T1D rows report `L1`, while Combined and AI-READI rows report `L2`
- the `K=9` block is naive-only and is included as an additional high-resolution baseline
- DE losses are evaluated at the exact notebook thresholds so they match the original real-data workflows

```python
from pathlib import Path

from IPython.display import Markdown, display

from tools.real_loss_tools import (
    compute_real_loss_tables,
    export_real_loss_tables,
    style_loss_comparison_table,
)

REPO_ROOT = Path.cwd()
```

## Compute naive thresholds and manuscript-aligned loss summaries

Naive thresholds are defined from the pooled empirical glucose distribution within each cohort:

- `K=2`: tertiles at probabilities `1/3` and `2/3`
- `K=4`: quintiles at probabilities `1/5`, `2/5`, `3/5`, and `4/5`
- `K=9`: deciles at probabilities `1/10`, `2/10`, `\ldots`, `9/10`

The reported loss table follows the manuscript's main-analysis choices: `L1` for Healthy and T1D, `L2` for Combined and AI-READI.

```python
tables = compute_real_loss_tables(REPO_ROOT)

threshold_summary_df = tables['threshold_summary_df']
loss_comparison_df = tables['loss_comparison_df']
naive_thresholds_wide_df = tables['naive_thresholds_wide_df']
reported_loss_wide_df = tables['reported_loss_wide_df']

display(Markdown('#### Pooled-quantile naive thresholds'))
display(naive_thresholds_wide_df)

display(Markdown('#### Main-experiment loss comparison'))
display(style_loss_comparison_table(reported_loss_wide_df))
```
